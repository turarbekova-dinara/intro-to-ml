## Geometric of Vectors

```python
v2 = [5, -3]
plt.plot([0, v2[0]], [0, v2[1]])

plt.axis((-6,6,-6,6))

plt.plot([-6,6], [0,0], 'b--')
plt.plot([0,0], [-6,6], 'r--')

plt.grid()
plt.show()
```

```python
v3 = [5,-4,3]

ax = plt.axes(projection = '3d')
ax.plot([0, v3[0]], [0, v3[1]], [0, v3[2]], linewidth = 3)

ax.plot([-5,5], [0,0], [0,0], 'r--')
ax.plot([0,0], [-5,5], [0,0], 'b--')
ax.plot([0,0], [0,0], [-6,6], 'g--')
plt.show()
```

```python
v1 = np.array([4, -2])
v2 = np.array([3,5])
v3 = v1 + v2
print("V3", v3)
plt.plot([0, v1[0]], [0, v1[1]], 'b', label = 'v1')
plt.plot([0, v2[0]] + v1[0], [0, v2[1]] + v1[1], 'r', label = 'v3')
plt.plot([0, v3[0]], [0, v3[1]], 'g', label = 'v3')

plt.plot([-7,7], [0,0], 'k--')
plt.plot([0,0], [-7,7], 'k--')

plt.legend()
plt.axis('square')
plt.axis((-8,8,-8,8))
plt.grid()
plt.show()
```
