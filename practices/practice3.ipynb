## 1.1 Data Cleaning

### 1.1.1 Handling Missing values 

```python
# 1. Drop columns with too many missing values
df_clean = df.drop(columns=["deck"])
```

```python
# 2. Fill 'age' with median value (robust to outliers)

# 5 10 15 25 200 = 255
# median = 15
# mean = 51
median_age = df_clean["age"].median()

# df_clean["age"].fillna(median_age, inplace=True)
df_clean["age"] = df_clean["age"].fillna(median_age)
```

```python
# 3. Fill "embarked" and "embark_town" with mode

mode_embarked = df_clean["embarked"].mode()[0]
mode_embark_town = df_clean["embark_town"].mode()[0]

df_clean["embarked"] = df_clean["embarked"].fillna(mode_embarked)
df_clean["embark_town"] = df_clean["embark_town"].fillna(mode_embark_town)
```

```python
df_clean.info()
```

### 1.1.2 Data types and Categories

```python
# Convert to categorical for memory and plotting

categorical_cols = ['sex', 'embarked', 'class', 'who', 'adult_male', 'alone']
for col in categorical_cols:
    df_clean[col] = df_clean[col].astype("category")
```

### 1.1.3 Deduplicate and reduntant columns 

```python
# Remove duplicated values 
df_clean.drop_duplicates(inplace=True)
```

```python
# Remove reduntant columns
df_clean.drop(columns = ["alive", "pclass", "embark_town"], inplace=True,
             errors="ignore")
```

```python
df_clean.head()
```

### 1.1.4 Outlier Detection

```python
# Boxplot for fare and age
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.boxplot(x=df_clean["fare"])
plt.title("Fare Distribution")

plt.subplot(1,2,2)
sns.boxplot(x=df_clean["age"])
plt.title("Age Distribution")

plt.show()
```

```python
# Optional: Cap Extreme fares using IQR method
Q1 = df_clean["fare"].quantile(0.25)
Q3 = df_clean["fare"].quantile(0.75)

IQR = Q3 - Q1

upper_bound = Q3 + 1.5*IQR

print(upper_bound)
df_clean["fare_capped"] = df_clean["fare"].clip(upper=upper_bound)
```

## 1.2 Visualization

### 1.2.1 Univariate plots

```python
plt.figure(figsize=(14,4))

#Age Distribution
plt.subplot(1,3,1)
sns.histplot(df_clean["age"], kde = True, bins = 30)
plt.title("Age Distrib")

#Fare Distribution (capped)
plt.subplot(1,3,2)
sns.histplot(df_clean["fare_capped"], kde = True, bins = 30)
plt.title("fare_capped Distrib")

#Survival rate
plt.subplot(1,3,3)
sns.countplot(x = "survived", data = df_clean)
plt.title("Survival rate")
```

```python
plt.figure(figsize=(14,4))

#Fare Distribution (capped)
plt.subplot(1,2,1)
sns.histplot(df_clean["fare_capped"], kde = True, bins = 30)
plt.title("fare_capped Distrib")

#Fare Distribution 
plt.subplot(1,2,2)
sns.histplot(df_clean["fare"], kde = True, bins = 30)
plt.title("fare Distrib")
```

### 1.2.2 Categorical relationship

```python
plt.figure(figsize=(14, 5))

#Survival rate by sex
plt.subplot(1,2,1)
sns.countplot(x='sex', hue = "survived", data = df_clean)
plt.title("Survival rate by sex")

#Survival rate by class
plt.subplot(1,2,2)
sns.countplot(x='class', hue = "survived", data = df_clean)
plt.title("Survival rate by class")
```

### 1.2.3 Numerical Relationship

```python
# Age vs Fare, colored by survival rate
sns.scatterplot(x = "age", y = "fare_capped", hue = "survived", style = "sex",
               alpha = 0.7, data = df_clean)
plt.title("Age vs Fare, colored by survival rate and sex")
plt.show()
```

```python
# Age vs Fare, colored by survival rate
sns.scatterplot(x = "age", y = "fare", hue = "survived", style = "sex",
               alpha = 0.7, data = df_clean)
plt.title("Age vs Fare, colored by survival rate and sex")
plt.show()
```

```python
# Boxplots: Age by Class
sns.boxplot(x = "class", y = "age", hue = "survived", data = df_clean)
plt.title("Age Distribution by Class ans Survival")
plt.show()
```

### 1.2.4 Pairplot and Heatmap

```python
# Pairplot for key variables 
sns.pairplot(df_clean, vars = ["age", 'fare_capped', "sibsp", "parch"],
            hue="survived", diag_kind = "kde")
plt.show()
```

```python
# Correlation heatmap
plt.figure(figsize=(8,6))

numeric_cols = df_clean.select_dtypes(include=np.number).columns

sns.heatmap(df_clean[numeric_cols].corr(), annot=True, cmap = 'coolwarm',
           center = 0)

plt.title("Correlation Heatmap")
plt.show()
```
