## Q5. Given vector **a = [1,2,3,4,5,6]**, compute its Euclidean norm manually and using NumPy.

Hint: By "manually," I mean solving it step by step using only basic NumPy functions.

### Answer 5

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])

a_squared = a**2

sum_squared = np.sum(a_squared)

manual_norm = np.sqrt(sum_squared)

print("Manual Euclidean norm:", manual_norm)

numpy_norm = np.linalg.norm(a)
print("NumPy linalg norm:", numpy_norm)
```

## Q6. Generate a random vector of length 10 and normalize it. Print the norm before and after normalization

### Answer 6

```python
import numpy as np

v = np.random.rand(10)
print("random vector: ", v)

norm_before = np.linalg.norm(v)
print("befor: ", norm_before)

v_normalized = v / norm_before

norm_after = np.linalg.norm(v_normalized)
print("after: ", norm_after)
print("normalized vector: ", v_normalized)
```

## Q7. Compute the dot product of vectors

a = [1,2,3,4,5,6] & b = [0,-4,3,-3,6,5]

using:

1. Manual loop
2. NumPy `np.dot()`.

### Answer 7

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])
b = np.array([0, -4, 3, -3, 6, 5])

dot_manual = 0
for i in range(len(a)):
  dot_manual += a[i] * b[i]

print("Scalar product (Manual loop):", dot_manual)

dot_numpy = np.dot(a, b)
print("Scalar product (NumPy np.dot()):", dot_numpy)
```

## Q8. Verify the distributive property:

$$a \cdot (b + c) = a \cdot b + a \cdot c$$

Hint: Generate three random numbers

### Answer 8

```python
import numpy as np

a = np.random.rand(3)
b = np.random.rand(3)
c = np.random.rand(3)

lhs = np.dot(a, b + c)
rhs = np.dot(a, b) + np.dot(a, c)

print("vector a:", a)
print("vector b:", b)
print("vector c:", c)

print("       ")

print("left part:", lhs)
print("right part:", rhs)

is_equal = np.allclose(lhs, rhs)
print("Answer for is_equal? :", is_equal)
```

## Q9: Compute the Euclidean distance matrix between every point in group A and every point in group B.

```python
# Group A (4,2)
A = np.array([
    [ 2,  1],
    [ 5,  4],
    [ 1,  6],
    [ 7,  2]
])

# Group B (3,2)
B = np.array([
    [0, 0],
    [6, 3],
    [3, 5]
])
```

```python
A_n = A[:, np.newaxis, :] #(4,1,2)
B_n = B[np.newaxis, :, :] #(1,3,2)

dif = A_n - B_n

sq_dif = dif ** 2

sum_sq = np.sum(sq_dif, axis=2)

dist_matrix = np.sqrt(sum_sq)

print("Euclidean distance matrix:\n", dist_matrix)
print("Shape of distance matrix:", dist_matrix.shape)
```
