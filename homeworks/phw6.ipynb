## Question 3:
**Do the following in one plot:**

1. Plot the vector as an arrow or line segment from the origin `(0,0)` to `(5,-3)`.
2. Add dashed x- and y-axes similar to the one in the material provided (horizontal and vertical dashed lines through the origin).
3. Set the plot limits so the vector is clearly visible (choose reasonable limits).

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([5, -3]) 
plt.plot([0, v[0]], [0, v[1]], 'y')

plt.axis((-6, 6, -6, 6))

plt.plot([-6, 6], [0, 0], 'r--')
plt.plot([0, 0], [-6, 6], 'g--')

plt.grid()
plt.show()
```

## Question 4:

**Do the following in one plot:**

1. Create a 3D plot and draw the vector from the origin to `(5, -4, 3)`.
2. Label the axes as `x`, `y`, `z`.

```python
import numpy as np
import matplotlib.pyplot as plt

v3 = np.array([5, -4, 3])

ax = plt.axes(projection='3d')

ax.plot([0, 5], [0, -4], [0, 3], linewidth=4)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

ax.plot([-6, 6], [0, 0], [0, 0], 'r--')
ax.plot([0, 0], [-6, 6], [0, 0], 'y--')
ax.plot([0, 0], [0, 0], [-6, 6], 'p--')

plt.show()
```

**5.2. On the same 2D figure, plot** all three vectors `v1`, `v2`, and `v3` from the origin.

```python
import numpy as np
import matplotlib.pyplot as plt

v1 = np.array([4, -2])
v2 = np.array([3, 5])
v3 = v1 + v2 

plt.plot([0, v1[0]], [0, v1[1]], 'b', label='v1')
plt.plot([v1[0], v1[0] + v2[0]], [v1[1], v1[1] + v2[1]], 'r', label='v2')
plt.plot([0, v3[0]], [0, v3[1]], 'g', label='v1+v2')

plt.plot([-8, 8], [0, 0], 'k--')
plt.plot([0, 0], [-8, 8], 'k--')

plt.legend()

plt.axis('square')
plt.axis((-8, 8, -8, 8))
plt.grid()
plt.show()
```
